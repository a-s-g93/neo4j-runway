# Google Gemini Integration

This notebook covers the basics of using Google Gemini LLM in Runway.

```python
import os

from neo4j_runway.llm.gemini import GeminiDiscoveryLLM, GeminiDataModelingLLM
from neo4j_runway import Discovery, UserInput, GraphDataModeler
import pandas as pd
```

```python
# First, install the required dependency:
# pip install google-generativeai

# Set your API key (get one from https://makersuite.google.com/app/apikey):
# export GEMINI_API_KEY="your-api-key-here"
# or set it directly in the code below
```

## Load Data

```python
df = pd.read_csv("../data/countries/countries.csv")
```

```python
user_input = UserInput(general_description='This is data on different countries.',
                                 column_descriptions={
                                     'id': 'unique id for a country.',
                                    'name': 'the country name.',
                                    'phone_code': 'country area code.',
                                    'capital': 'the capital of the country.',
                                    'currency_name': "name of the country's currency.",
                                    'region': 'primary region of the country.',
                                    'subregion': 'subregion location of the country.',
                                    'timezones': 'timezones contained within the country borders.',
                                    'latitude': 'the latitude coordinate of the country center.',
                                    'longitude': 'the longitude coordinate of the country center.'
                                    },
                                use_cases=[
        "Which region contains the most subregions?", 
        "What currencies are most popular?", 
        "Which countries share timezones?"
    ]  
                                )
```

## Discovery

### About Gemini 2.5 Pro

**Gemini 2.5 Pro** is Google's latest and most advanced model, offering:
- **Enhanced reasoning capabilities** for better data discovery
- **Improved code generation** for data modeling tasks  
- **Better instruction following** for complex graph modeling
- **Cost-effective** compared to other enterprise models
- **Large context window** for handling complex datasets

Perfect for neo4j-runway's data discovery and graph modeling workflows!

```python
import os
from your_module import GeminiDiscoveryLLM, GeminiDataModelingLLM  # Adjust the import based on your project structure

# Get your API key from https://makersuite.google.com/app/apikey
api_key = os.environ.get("GEMINI_API_KEY")  

# Create discovery LLM
llm_disc = GeminiDiscoveryLLM(
    model_name='gemini-2.5-pro',  # Latest Gemini model
    model_params={"temperature": 0.5},
    api_key=api_key
)

# Create data modeling LLM
llm_dm = GeminiDataModelingLLM(
    model_name='gemini-2.5-pro',  # Latest Gemini model  
    model_params={"temperature": 0.5},
    api_key=api_key
)

print(f"✅ Discovery LLM: {llm_disc.model_name}")
print(f"✅ Data Modeling LLM: {llm_dm.model_name}")
```

```python
disc = Discovery(llm=llm_disc, user_input=user_input, data=df)
```

```python
disc.run(show_result=True, notebook=True)
```

## Data Modeling

```python
dm_llm = GeminiDataModelingLLM(model_name="gemini-1.5-pro", # or "gemini-1.5-flash" for faster responses
                               model_params={"temperature": 0.0},
                               api_key=os.environ.get("GEMINI_API_KEY"), # this will be auto read from env if present. declaring here for demo.
                               )

# Data modeling LLM is already created above as llm_dm
# Using the same gemini-2.5-pro model for consistency
print(f"Using {llm_dm.model_name} for data modeling with temperature {llm_dm.model_params.get('temperature', 'default')}")
```

```python
gdm = GraphDataModeler(discovery=disc, llm=llm_dm)
```

```python
gdm.create_initial_model(max_retries=2)
```

```python
gdm.current_model.visualize()
```
